$$
MSE = E = \frac{\sum_{i=0}^n (y_{i} - (m \cdot x_{i} + b))^2}{n}
$$

$$
\begin{aligned}
\frac{\partial E}{\partial m} 
&= \frac{\sum_{i=0}^n 2 \cdot (y_{i} - (m \cdot x_{i} + b)) \cdot (-x_{i})}{n} \\
&= \frac{-2}{n} \cdot \sum_{i=0}^n x_{i} (y_{i} - (m \cdot x_{i} + b))
\end{aligned}
$$

$$
\begin{aligned}
\frac{\partial E}{\partial b} = \frac{-2}{n} \cdot \sum_{i=0}^n (y_{i} - (m \cdot x_{i} + b))
\end{aligned}
$$

$$
m = m - LR \cdot \frac{\partial E}{\partial m} \\
b = b - LR \cdot \frac{\partial E}{\partial b} \\
LR = 0.001
$$

$$
h(X(i)) = \sigma(Z(i))

$$
L(\theta) = \prod_{i=1}^{m} [h(X(i))]^{y(i)} \cdot [1 - h(X(i))]^{1-y(i)}
$$

$$
\ln(L(\theta)) = l = \sum_{i=1}^{m} y(i) \cdot \ln(h(X(i))) + (1-y(i)) \cdot \ln(1-h(X(i))) \\
\begin{aligned}
\zeta(\theta) 
&= \frac{-1}{m} \sum_{i=1}^{m} y(i) \cdot \ln(h(X(i))) + (1-y(i)) \cdot \ln(1-h(X(i))) \\
&= \frac{-l}{m}
\end{aligned}
$$

$$
\frac{\partial J}{\partial \theta(j)} \\

\begin{aligned}
\frac{\partial l}{\partial \theta(j)} 
&= \frac{\partial l}{\partial h} \cdot \frac{\partial h}{\partial z} \cdot \frac{\partial z}{\partial \theta(j)} \\
&= [\frac{y(i)}{h(i)} - \frac{1-y(i)}{1-h(i)}] \cdot [h(i)(1-h(i))] \cdot x(i)_{j} \\
&= (y(i) - h(i)) \cdot x(i)_{j}
\end{aligned}
$$

$$
J = \frac{-l}{m} \cdot l(\theta) \\
\frac{\partial J}{\partial \theta(j)} = \frac{-l}{m} \cdot (y(i) - h(i)) \cdot x(i)_{j}
